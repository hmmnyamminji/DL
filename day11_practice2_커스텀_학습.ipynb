```python
# 셀 1. 전이학습
model = YOLO("yolo11n.pt")        # pretrained 가중치 코드 = 전이학습의 출발점, 특징 호출
```

```python
# 셀 3. 연습 학습 실험 - coco8로 파이프라인 체험
results = model.train( # 재학습 실험,
    data="coco8.yaml",        # data.yaml 학습 주문서
    epochs=3,
    imgsz=320,
    batch=4,                # 한 번에 4장씩
    project="runs_practice",  # 결과 상위 폴더명
    name="coco8_test",        # 이번 실행 이름
    exist_ok=True,            # 덮어쓰기 허용
    verbose=False,            # 진행 로그 최소화

)
```

```python
# 셀 4. runs/ 폴더 해부 - 만들어진 것들
import os
save_dir = str(results.save_dir)
print("\n[결과 폴더 내용물]")
for f in sorted(os.listdir(save_dir)): # listdir - 폴더 안 파일 목록, 정렬해서 출력
  print(" ", f)

print("   weights/ ->", os.listdir(os.path.join(save_dir, "weights")))
```
